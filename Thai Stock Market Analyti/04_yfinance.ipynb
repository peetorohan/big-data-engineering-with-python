```python
#แปลงข้อมูลให้พร้อมใช้
all_data = []

for t in tickers_yf:
    try:
        df = data[t].copy()
        df["symbol"] = t.replace(".BK", "")
        df = df.reset_index()
        all_data.append(df)
    except:
        print(f"❌ ไม่มีข้อมูล: {t}")

final_df = pd.concat(all_data)
final_df.columns = [c.lower().replace(" ", "_") for c in final_df.columns]

final_df.head()
```

```python
#map เป็น stock_id
stocks_df = pd.read_sql("SELECT stock_id, symbol FROM stocks", engine)

stock_map = dict(zip(stocks_df["symbol"], stocks_df["stock_id"]))

final_df["stock_id"] = final_df["symbol"].map(stock_map)
```

```python
#ปรับ column ให้ตรง DB
final_df.rename(columns={
    "date": "trade_date"
}, inplace=True)

final_df = final_df[
    ["trade_date", "open", "high", "low", "close", "volume", "stock_id"]
]
```

```python
import yfinance as yf
import pandas as pd

data = []

for sym in symbols:
    try:
        ticker = yf.Ticker(sym + ".BK")  # หุ้นไทยต้องเติม .BK
        info = ticker.info
        name = info.get("longName") or info.get("shortName") or sym
        
        data.append({
            "symbol": sym,
            "stock_name": name
        })
    except:
        data.append({
            "symbol": sym,
            "stock_name": sym
        })

df = pd.DataFrame(data)
df
```

```python
from sqlalchemy import text

with engine.begin() as conn:
    for _, row in df.iterrows():
        conn.execute(text("""
            INSERT INTO stocks (symbol, stock_name)
            VALUES (:symbol, :stock_name)
            ON CONFLICT (symbol)
            DO UPDATE SET
                stock_name = EXCLUDED.stock_name
        """), {
            "symbol": row["symbol"],
            "stock_name": row["stock_name"]
        })

print("✅ Upsert completed (no duplicate error)")
```

```python
def map_category(sym):
    s = sym.upper()

    # REIT & PF
    if "REIT" in s or "PF" in s or "IFF" in s:
        return 19

    # Banking
    if s in ["KBANK","SCB","KTB","TTB","BBL","KKP","TISCO"]:
        return 4

    # Finance
    if s in ["MTC","SAWAD","KTC","AEONTS","JMT","BAM"]:
        return 5

    # Energy
    if s.startswith(("PTT","PTTEP","PTTGC","GULF","GPSC","EA","EGCO","RATCH","SPRC","TOP")):
        return 3

    # ICT
    if s.startswith(("ADVANC","TRUE","DTAC","INTUCH","INET","IT","JAS","THCOM","ILINK")):
        return 6

    # Commerce
    if s.startswith(("CPALL","HMPRO","GLOBAL","DOHOME","CRC","COM7","BJC","MAKRO")):
        return 7

    # Property
    if s.startswith(("LH","ORI","SPALI","SIRI","QH","NOBLE","SC","LPN","ANAN")):
        return 8

    # Construction Services
    if s.startswith(("ITD","STEC","UNIQ","SEAFCO","PYLON","CK")):
        return 9

    # Construction Materials
    if s.startswith(("SCC","TPIPL","TOA","TASCO")):
        return 10

    # Transport
    if s.startswith(("AOT","BEM","BTS","RCL","PSL","WICE","SJWD","PORT")):
        return 11

    # Tourism
    if s.startswith(("MINT","CENTEL","ERW","DUSIT","SPA")):
        return 12

    # Health
    if s.startswith(("BDMS","BH","CHG","RAM","VIH","EKH")):
        return 13

    # Media
    if s.startswith(("WORK","MONO","GRAMMY","RS","ONEE")):
        return 14

    # Packaging
    if s.startswith(("SCGP","BGC")):
        return 15

    # Industrial
    if s.startswith(("DELTA","HANA","KCE","SVI","SMT")):
        return 16

    # Automotive
    if s.startswith(("SAT","AH","STANLY")):
        return 17

    # Insurance
    if s.startswith(("TIPH","THRE","TVH","BLA")):
        return 18

    # Food
    if s.startswith(("CPF","TU","GFPT","OSP","ICHI","SAPPE","TKN")):
        return 2

    # Agro
    if s.startswith(("STA","NER","TRUBB")):
        return 1

    # Default
    return 20
```

```python
import pandas as pd

df = pd.DataFrame(symbols, columns=["symbol"])
df["category_id"] = df["symbol"].apply(map_category)

df.head()
```

```python
from psycopg2.extras import execute_batch
import psycopg2

conn = psycopg2.connect(
    host="localhost",
    database="thai_stock_market",
    user="postgres",
    password="admin"
)
cur = conn.cursor()

data = list(df.itertuples(index=False, name=None))

execute_batch(cur, """
UPDATE stocks
SET category_id = %s
WHERE symbol = %s
""", [(cat, sym) for sym, cat in data])

conn.commit()
cur.close()
conn.close()
```
